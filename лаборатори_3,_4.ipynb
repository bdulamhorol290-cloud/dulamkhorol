###1.Массиваас анхны тоонуудыг хэвлэх

```python
import random

n = int(input("n="))
a = [random.randint(20, 100) for _ in range(n)]

print(a)

for x in a:
    if all(x % i != 0 for i in range(2, x)) and x > 1:
        print(x, end=" ")
```

###2. Палиндром тоонууд

```python
import random

n = int(input("n="))
a = [random.randint(20, 100) for _ in range(n)]

print(a)

for x in a:
    if str(x) == str(x)[::-1]:
        print(x, end=" ")
```

###7. Анхны тоонуудаар массив үүсгэх

```python
n = int(input("n="))

a = []
x = 2

while len(a) < n:
    p = True
    for i in range(2, x):
        if x % i == 0:
            p = False
            break
    if p:
        a.append(x)
    x += 1

print(a)
```

### 8. Палиндром тоонуудаар массив үүсгэх

```python
n = int(input("n="))

a = []
x = 10

while len(a) < n:
    if str(x) == str(x)[::-1]:
        a.append(x)
    x += 1

print(a)
```

###9. 112-т хуваахад 111 үлддэг тоо хэд байгааг тоолох

```python
import random

n = int(input("n="))

a = [random.randint(20, 1000) for i in range(n)]

print(a)

c = 0
for x in a:
    if x % 112 == 111:
        c += 1

print("Тоо =", c)
```

###10. a[n] массивын элементүүдийг хөрвүүлж b[n] массив үүсгэх

```python
import random

n = int(input("n="))

a = [random.randint(20, 100) for i in range(n)]

b = []

for x in a:
    b.append(int(str(x)[::-1]))

print("a =", a)
print("b =", b)
```

###Дүнгийн массив өгөгдсөн бол хамгийн сайн ба муу дүнг ол.

```python
import random

n = int(input("n="))
A = [random.randint(0, 100) for i in range(n)]

print("Дүн:", A)

mx, mn = max(A), min(A)

print("Хамгийн сайн дүн:", mx)
print("Хамгийн муу дүн:", mn)
print("Зөрүү:", mx - mn)

print("Сайн дүнтэй оюутны тоо:", A.count(mx))
print("Муу дүнтэй оюутны тоо:", A.count(mn))

print("Сайн дүнтэй оюутнуудын дугаар:",
      [i + 1 for i in range(n) if A[i] == mx])

print("Муу дүнтэй оюутнуудын дугаар:",
      [i + 1 for i in range(n) if A[i] == mn])
```
